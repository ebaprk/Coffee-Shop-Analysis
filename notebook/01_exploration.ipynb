# Coffee Shop Sales Analysis

## Milestone 1: Data Exploration

Objective: Understand the dataset, inspect its
structure, and identify potential data quality issues.

```python
import pandas as pd

from pathlib import Path

# Find the workbook from the project root or any folder within the project.
project_root = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents)
     if (folder / "data" / "Coffee Shop Sales.xlsx").is_file()),
    None
)
if project_root is None:
    raise FileNotFoundError(
        "Open the notebook from the Coffee Shop Analysis project folder "
        "or a folder inside it. Expected data/Coffee Shop Sales.xlsx."
    )
df = pd.read_excel(project_root / "data" / "Coffee Shop Sales.xlsx")
```

```python
# 1. Display the first 10 rows of the dataset.
display(df.head(10))

# 2. Determine how many rows and columns it contains.
display(df.shape)

# 3. Identify the data type of every column.
display(df.dtypes)

# 4. Check whether any columns contain missing values.
display(df.isna().sum())

# 5. Check for duplicate rows
display(df.duplicated().sum())

# 6.
display("Earliest transaction: ", df["transaction_date"].min())
display("Latest transaction: ", df["transaction_date"].max())
```

## Milestone 1, Challenge 02: Data Quality Validation

Check for missing values, invalid values, duplicate IDs, inconsistent category labels, and invalid dates and times.

```python
# Task 1: Check for invalid quantities
invalid_quantities = (df["transaction_qty"] <= 0).sum()
display(invalid_quantities)

# Task 2: Check for invalid prices
invalid_prices = (df["unit_price"] <= 0).sum()
display(invalid_prices)

# Task 3: Investigate transaction IDs
duplicate_ids = df["transaction_id"].duplicated().sum()
display(duplicate_ids)

# Task 4: Check store and product category label consistency
# These counts find extra labels caused by capitalization or outer spaces.
original_stores = df["store_location"].nunique()
cleaned_stores = (
    df["store_location"]
    .str.strip()
    .str.casefold()
    .nunique()
)
store_issues = original_stores - cleaned_stores

original_categories = df["product_category"].nunique()
cleaned_categories = (
    df["product_category"]
    .str.strip()
    .str.casefold()
    .nunique()
)
category_issues = original_categories - cleaned_categories

display(pd.Series({
    "Store naming variants": store_issues,
    "Product category naming variants": category_issues
}))

# Task 5: Validate dates and times
display(df["transaction_time"].head())
display(df["transaction_time"].map(type).value_counts())
display(df["transaction_date"].head())
display(df["transaction_date"].map(type).value_counts())

parsed_dates = pd.to_datetime(
    df["transaction_date"],
    errors="coerce"
)
invalid_dates = (
    parsed_dates.isna() &
    df["transaction_date"].notna()
).sum()

parsed_times = pd.to_datetime(
    df["transaction_time"].astype(str),
    format="%H:%M:%S",
    errors="coerce"
)
invalid_times = (
    parsed_times.isna() &
    df["transaction_time"].notna()
).sum()

# Task 6: Summarize data quality issues
missing_rows = df.isna().any(axis=1).sum()

quality_summary = pd.DataFrame([
    {
        "Check": "Invalid quantities",
        "Issues Found": invalid_quantities,
        "Unit": "rows",
        "Issue Detected": invalid_quantities > 0
    },
    {
        "Check": "Invalid prices",
        "Issues Found": invalid_prices,
        "Unit": "rows",
        "Issue Detected": invalid_prices > 0
    },
    {
        "Check": "Duplicated IDs",
        "Issues Found": duplicate_ids,
        "Unit": "rows",
        "Issue Detected": duplicate_ids > 0
    },
    {
        "Check": "Missing values",
        "Issues Found": missing_rows,
        "Unit": "rows",
        "Issue Detected": missing_rows > 0
    },
    {
        "Check": "Store naming variants",
        "Issues Found": store_issues,
        "Unit": "labels",
        "Issue Detected": store_issues > 0
    },
    {
        "Check": "Product category naming variants",
        "Issues Found": category_issues,
        "Unit": "labels",
        "Issue Detected": category_issues > 0
    },
    {
        "Check": "Invalid dates",
        "Issues Found": invalid_dates,
        "Unit": "rows",
        "Issue Detected": invalid_dates > 0
    },
    {
        "Check": "Invalid times",
        "Issues Found": invalid_times,
        "Unit": "rows",
        "Issue Detected": invalid_times > 0
    }
])
display(quality_summary)
```
